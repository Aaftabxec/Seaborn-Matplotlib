```python
# Import libraries for plotting, statistical visualization, and data handling.
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
# Display the datasets available in Seaborn so we can choose one to analyze.
sns.get_dataset_names()
```

```python
# Load the diamonds dataset from Seaborn into a pandas DataFrame.
df= sns.load_dataset("diamonds")
```

```python
# Display the DataFrame to inspect the dataset rows and columns.
df
```

```python
# Show column names, data types, non-null counts, and memory usage.
df.info()
```

```python
# bar plot for a value and category
```

```python
sns.barplot(data=df, x='cut', y='price')
plt.title("Average Diamond Price by Caytegories")
plt.xlabel("Diamond Cut")
plt.ylabel("Average Price")
plt.show()
```

```python
#histplot for a numerical diststrinution...Group numerical values into ranges and count them.
```

```python
# Create a histogram to show how diamond prices are distributed across 5 bins.
sns.histplot(df['price'], bins=5)

plt.show()   
```

```python
# A scatter plot helps us understand the relationship between two numerical variables.
```

```python
sns.scatterplot(data=df, x='carat', y='price')

plt.show()
```

```python
# A box plot helps us understand the spread of data..
```

```python
sns.boxplot(data=df, x='cut', y='price')

plt.show()
```

```python
# A violin plot shows how data is distributed and where most values are concentrated.
```

```python
sns.violinplot(data=df,x='cut',y='price')

plt.show()
```

```python
# Calculate correlations between selected numeric columns and display them as a heatmap.
sns.heatmap(df[['carat', 'depth', 'table', 'price', 'x', 'y', 'z']].corr(), annot=True)
plt.show()
```
